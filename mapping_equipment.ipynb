# Equipment stats for outages in outlier regions

So for each outlier tract, I am trying to see if there are any equipment that are repeat offenders (drive outages often). So for each tract, creat a sub data frame with all the outages related to it in outage_events (match on TRACT). And then find all the unique device_lat and device_lon (two columns) combinations. A table that shows frequency of each outage and all the related stats: total number that device has seen outages, total time out (fix_duration_hours column) total and average customer_minutes_out, to get idea of frequency and duration of each device being out.

```python
outage_events = pd.read_csv('data_processed/PCA_data/outage_events.csv')

tract_groups = {
    "ranked_most": ranked_most,
    "study_tracts": study_tracts,
    "rtp_focus": rtp_focus,
    "charlotte_focus": charlotte_focus,
    "wnc_focus": wnc_focus,
}

tract_lookup = (
    pd.DataFrame(
        [(tract, group) for group, tracts in tract_groups.items() for tract in tracts],
        columns=["TRACT", "tract_group"]
    )
    .drop_duplicates()
)
```

```python
outage_events["TRACT"] = outage_events["TRACT"].astype(str)
tract_lookup["TRACT"] = tract_lookup["TRACT"].astype(str)
```

```python
outage_focus = outage_events.merge(
    tract_lookup,
    on="TRACT",
    how="inner"
)

device_cols = ["device_lat", "device_lon"]

device_summary = (
    outage_focus
    .groupby(["TRACT", "tract_group"] + device_cols)
    .agg(
        outage_count=("outage_identifer", "nunique"),

        total_fix_hours=("fix_duration_hours", "sum"),
        avg_fix_hours=("fix_duration_hours", "mean"),
        max_fix_hours=("fix_duration_hours", "max"),

        total_household_min_out=("customer_minutes_out", "sum"),
        avg_household_min_out=("customer_minutes_out", "mean"),
        max_household_min_out=("customer_minutes_out", "max"),

        avg_household_affected=("affected", "mean"),
        max_household_affected=("affected", "max"),

        summer_outages=("is_summer", "sum"),
        winter_outages=("is_winter", "sum"),

        helene_outages=("overlaps_helene", "sum"),

        first_outage=("outage_start_estimate", "min"),
        last_outage=("outage_end_estimate", "max"),
    )
    .reset_index()
)
```

```python
device_summary["first_outage"] = pd.to_datetime(device_summary["first_outage"], errors="coerce")
device_summary["last_outage"]  = pd.to_datetime(device_summary["last_outage"],  errors="coerce")

#Outages per year
device_summary["outages_per_year"] = (
    device_summary["outage_count"] /
    ((device_summary["last_outage"] - device_summary["first_outage"]).dt.days / 365)
)

#share of outages within tract
device_summary["share_of_tract_outages"] = (
    device_summary["outage_count"] /
    device_summary.groupby("TRACT")["outage_count"].transform("sum")
)

#severity per outage
device_summary["household_min_per_outage"] = (
    device_summary["total_household_min_out"] /
    device_summary["outage_count"]
)

#rank worst devices in each tract
device_summary["repeat_offender_score"] = (
    device_summary["outage_count"] *
    device_summary["household_min_per_outage"] *
    device_summary["avg_household_affected"]
)

device_summary = device_summary.sort_values(
    ["TRACT", "repeat_offender_score"],
    ascending=[True, False]
)

```

```python
# =====================================================
# 1) GET TOP 10 DEVICES PER TRACT
# =====================================================

topk = 10

top_devices_by_tract = (
    device_summary.sort_values(["TRACT", "repeat_offender_score"], ascending=[True, False])
    .groupby("TRACT")
    .head(topk)
    .reset_index(drop=True)
)

top_devices_by_tract["rank_in_tract"] = (
    top_devices_by_tract.groupby("TRACT")["repeat_offender_score"]
    .rank(method="first", ascending=False)
    .astype(int)
)

# =====================================================
# 2) CONVERT TO GEODATAFRAME
# =====================================================

dev_pts_gdf = gpd.GeoDataFrame(
    top_devices_by_tract,
    geometry=gpd.points_from_xy(
        top_devices_by_tract["device_lon"],
        top_devices_by_tract["device_lat"]
    ),
    crs="EPSG:4326"
)

dev_pts_gdf["lon"] = dev_pts_gdf.geometry.x
dev_pts_gdf["lat"] = dev_pts_gdf.geometry.y

# =====================================================
# 3) PREPARE TRACT GROUP LABELS
# =====================================================

tract_geo = tract_geo.copy()

bool_cols = []
for name, geoids in groups.items():
    if name not in tract_geo.columns:
        tract_geo[name] = tract_geo["GEOID"].isin(geoids)
    bool_cols.append(name)

def combined_label(row):
    present = [c for c in bool_cols if row[c]]
    if not present:
        return "Other"
    if len(present) == 1:
        return present[0]
    return " & ".join(present)

tract_geo["plot_label"] = tract_geo.apply(combined_label, axis=1)

# =====================================================
# 4) COLOR MAP
# =====================================================

base_colors = {
    "Ranked across most columns": "#d73027",
    "Study tracts (all data)": "#1a9850",
    "RTP focus": "#4575b4",
    "Charlotte focus": "#fdae61",
    "WNC focus": "#7b3294"
}

unique_labels = sorted(tract_geo["plot_label"].unique())
color_map = {}

for label in unique_labels:
    if label in base_colors:
        color_map[label] = base_colors[label]
    elif label == "Other":
        color_map[label] = "#f0f0f0"
    else:
        parts = [p.strip() for p in label.split("&")]
        rgbs = []
        for p in parts:
            if p in base_colors:
                hexc = base_colors[p].lstrip("#")
                rgbs.append(tuple(int(hexc[i:i+2], 16) for i in (0,2,4)))
        if rgbs:
            avg = tuple(int(sum(c[i] for c in rgbs)/len(rgbs)) for i in range(3))
            color_map[label] = "#{:02x}{:02x}{:02x}".format(*avg)
        else:
            color_map[label] = "#999999"

# =====================================================
# 5) CONVERT TRACTS TO EPSG:4326
# =====================================================

tract_geo_4326 = tract_geo.to_crs(epsg=4326).copy()

features_by_label = {}

for label in unique_labels:
    subset = tract_geo_4326[tract_geo_4326["plot_label"] == label]
    if subset.empty:
        features_by_label[label] = None
        continue

    feats = []
    for _, row in subset.iterrows():
        feats.append({
            "type": "Feature",
            "geometry": mapping(row.geometry),
            "properties": {"GEOID": row["GEOID"]}
        })

    features_by_label[label] = {
        "type": "FeatureCollection",
        "features": feats
    }

# =====================================================
# 6) DEVICE HOVER TEXT
# =====================================================

def mk_hover(row):
    return (
        f"<b>TRACT</b>: {row['TRACT']}<br>"
        f"<b>Rank</b>: {row['rank_in_tract']} / {topk}<br>"
        f"<b>Score</b>: {row['repeat_offender_score']:,.0f}<br>"
        f"<b>Outages</b>: {row.get('outage_count', 0)}<br>"
        f"<b>Households</b>: {row.get('households_impacted', 0)}"
    )

dev_pts_gdf["hover_text"] = dev_pts_gdf.apply(mk_hover, axis=1)

# =====================================================
# 7) BUILD MAP WITH DROPDOWN FILTER
# =====================================================

center = tract_geo_4326.geometry.unary_union.centroid
center_lon, center_lat = center.x, center.y

fig = go.Figure()

trace_map = {}  # track which traces belong to which label

# -----------------------------------------------------
# ADD TRACES (tracts + devices per label)
# -----------------------------------------------------

for label in unique_labels:
    if label == "Other":
        continue

    trace_map[label] = []

    # --- TRACTS ---
    geojson = features_by_label[label]
    if geojson is not None:
        locs = [f["properties"]["GEOID"] for f in geojson["features"]]

        fig.add_trace(go.Choroplethmapbox(
            geojson=geojson,
            locations=locs,
            z=[1]*len(locs),
            featureidkey="properties.GEOID",
            colorscale=[[0, color_map[label]], [1, color_map[label]]],
            showscale=False,
            marker_line_width=0.6,
            marker_line_color="black",
            marker_opacity=0.42,
            name=f"{label} tracts",
            visible=True
        ))
        trace_map[label].append(len(fig.data)-1)

    # --- DEVICES ---
    geoids = set(
        tract_geo_4326.loc[
            tract_geo_4326["plot_label"] == label, "GEOID"
        ]
    )

    subset = dev_pts_gdf[dev_pts_gdf["TRACT"].isin(geoids)]

    if not subset.empty:
        sizes = (np.sqrt(subset["repeat_offender_score"] + 1) * 1.5).clip(3, 12)

        fig.add_trace(go.Scattermapbox(
            lon=subset["lon"],
            lat=subset["lat"],
            mode="markers",
            marker=dict(size=sizes, color="red", opacity=0.9),
            text=subset["hover_text"],
            hoverinfo="text",
            name=f"{label} devices",
            visible=True
        ))
        trace_map[label].append(len(fig.data)-1)

# -----------------------------------------------------
# CREATE DROPDOWN BUTTONS
# -----------------------------------------------------

buttons = []

# --- Show ALL ---
all_visible = [True] * len(fig.data)

buttons.append(dict(
    label="Show All",
    method="update",
    args=[{"visible": all_visible}]
))

# --- Individual labels ---
for label in unique_labels:
    if label == "Other":
        continue

    visible = [False] * len(fig.data)

    for idx in trace_map[label]:
        visible[idx] = True

    buttons.append(dict(
        label=label,
        method="update",
        args=[{"visible": visible}]
    ))

# -----------------------------------------------------
# LAYOUT
# -----------------------------------------------------

fig.update_layout(
    title=f"Top {topk} Repeat-Offender Devices per Census Tract",
    margin={"r":0,"t":60,"l":0,"b":0},
    mapbox=dict(
        style="carto-positron",
        center={"lat": center_lat, "lon": center_lon},
        zoom=8.5
    ),
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            showactive=True,
            x=0.02,
            xanchor="left",
            y=0.98,
            yanchor="top"
        )
    ]
)

# =====================================================
# 8) SAVE
# =====================================================

out_html = f"top_{topk}_repeat_offenders_map.html"
fig.write_html(out_html, include_plotlyjs="cdn")

print(f"Saved interactive map to {out_html}")
```

```python
import plotly.graph_objects as go
import numpy as np

fig = go.Figure()

tract_ids = sorted(dev_pts_gdf["TRACT"].unique())

trace_lookup = {}  # store trace indices per tract

# =====================================================
# ADD TRACES PER TRACT
# =====================================================

for tract in tract_ids:

    # --- All devices in this tract (for heatmap)
    all_devices = device_summary[device_summary["TRACT"] == tract]

    heat_trace = go.Densitymapbox(
        lon=all_devices["device_lon"],
        lat=all_devices["device_lat"],
        z=np.sqrt(all_devices["repeat_offender_score"]),  # smooth weighting
        radius=40,  # MUCH larger smoothing
        colorscale="YlOrRd",
        opacity=0.6,
        visible=False,
        name=f"{tract} Problem Density"
    )
    fig.add_trace(heat_trace)
    heat_index = len(fig.data) - 1

    # --- Top 10 devices in this tract
    top_devices = dev_pts_gdf[dev_pts_gdf["TRACT"] == tract]

    sizes = (np.sqrt(top_devices["repeat_offender_score"] + 1) * 1.5).clip(3, 10)

    top_trace = go.Scattermapbox(
        lon=top_devices["lon"],
        lat=top_devices["lat"],
        mode="markers",
        marker=dict(size=sizes, opacity=0.9),
        text=top_devices["hover_text"],
        hoverinfo="text",
        visible=False,
        name=f"{tract} Top 10"
    )
    fig.add_trace(top_trace)
    top_index = len(fig.data) - 1

    trace_lookup[tract] = {
        "heat": heat_index,
        "top": top_index
    }

# =====================================================
# DROPDOWN MENU FOR TRACT SELECTION
# =====================================================

buttons = []

for tract in tract_ids:

    visibility = [False] * len(fig.data)

    # Turn on both layers for selected tract
    visibility[trace_lookup[tract]["heat"]] = True
    visibility[trace_lookup[tract]["top"]] = True

    buttons.append(
        dict(
            label=f"TRACT {tract}",
            method="update",
            args=[{"visible": visibility}]
        )
    )

fig.update_layout(
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            showactive=True,
            x=0.01,
            y=0.98
        )
    ]
)

# =====================================================
# LAYOUT
# =====================================================

center = tract_geo_4326.geometry.unary_union.centroid

fig.update_layout(
    mapbox_style="open-street-map",
    mapbox_zoom=9,
    mapbox_center={"lat": center.y, "lon": center.x},
    margin={"r":0,"t":40,"l":0,"b":0},
    title="Interactive Tract-Level Equipment Analysis"
)

fig.show()
```

From outage_events, I want to calculate a repeat_offender_score for each device_lat, device_lon combination. And then create a continuous surface across all of north carolina to see if there are hot spots of devices.

## Compute Device Level Scores

```python
import pandas as pd

# Clean columns just in case
outage_events.columns = outage_events.columns.str.strip()

# Drop rows missing device coordinates
events_clean = outage_events.dropna(subset=["device_lat", "device_lon"])

# Aggregate to device level
device_scores = (
    events_clean
    .groupby(["device_lat", "device_lon"], as_index=False)
    .agg(
        outage_count=("outage_identifer", "nunique"),
        total_customer_minutes=("customer_minutes_out", "sum"),
        avg_fix_hours=("fix_duration_hours", "mean")
    )
)

# Define repeat offender score
device_scores["repeat_offender_score"] = device_scores["total_customer_minutes"]

print(device_scores.head())
```

## Mapping

```python
#Create continuous statewide surface
device_gdf = gpd.GeoDataFrame(
    device_scores,
    geometry=gpd.points_from_xy(
        device_scores["device_lon"],
        device_scores["device_lat"]
    ),
    crs="EPSG:4326"
)
```

```python
#Statewide heat map
import plotly.graph_objects as go
import numpy as np

fig = go.Figure()

fig.add_trace(go.Densitymapbox(
    lon=device_gdf["device_lon"],
    lat=device_gdf["device_lat"],
    z=np.sqrt(device_gdf["repeat_offender_score"]),  # prevents domination by extreme values
    radius=20,  # adjust smoothing (20–50 good range)
    colorscale="YlOrRd",
    opacity=0.9,
    name="Device Hot Spots"
))

# Center roughly on NC
fig.update_layout(
    mapbox_style="open-street-map",
    mapbox_zoom=6.5,
    mapbox_center={"lat": 35.5, "lon": -79.0},
    margin={"r":0,"t":40,"l":0,"b":0},
    title="Statewide Device Hot Spot Surface (Customer Minutes Weighted)"
)

fig.show()
```

```python
import h3
import numpy as np
import plotly.graph_objects as go

# -------------------------------------------------
# 1️⃣ Choose hex resolution
# 6 = very large
# 7 = statewide clean
# 8 = more detailed
# -------------------------------------------------
resolution = 7

# -------------------------------------------------
# 2️⃣ Assign each device to a hex (FAST version)
# -------------------------------------------------
device_scores["h3_index"] = [
    h3.latlng_to_cell(lat, lon, resolution)
    for lat, lon in zip(device_scores["device_lat"], device_scores["device_lon"])
]

# -------------------------------------------------
# 3️⃣ Aggregate score per hex
# -------------------------------------------------
hex_agg = (
    device_scores
    .groupby("h3_index", as_index=False)
    .agg(total_score=("repeat_offender_score", "sum"))
)

# Optional: log transform for better contrast
hex_agg["score_scaled"] = np.log1p(hex_agg["total_score"])

# -------------------------------------------------
# 4️⃣ Convert hex cells to GeoJSON polygons
# -------------------------------------------------
features = []

for _, row in hex_agg.iterrows():

    boundary = h3.cell_to_boundary(row["h3_index"])

    # Convert (lat, lon) → (lon, lat)
    boundary_lonlat = [(lng, lat) for lat, lng in boundary]

    features.append({
        "type": "Feature",
        "geometry": {
            "type": "Polygon",
            "coordinates": [boundary_lonlat]
        },
        "properties": {
            "h3_index": row["h3_index"]
        }
    })

geojson = {
    "type": "FeatureCollection",
    "features": features
}

# -------------------------------------------------
# 5️⃣ Build Plotly Choropleth
# -------------------------------------------------
fig = go.Figure(go.Choroplethmapbox(
    geojson=geojson,
    locations=hex_agg["h3_index"],
    z=hex_agg["score_scaled"],
    featureidkey="properties.h3_index",
    colorscale="YlOrRd",
    marker=dict(
        line=dict(width=0.3, color="white"),
        opacity=0.6  
    ),
    colorbar_title="Hot Spot Intensity"
))

fig.update_layout(
    mapbox_style="open-street-map",
    mapbox_zoom=6.5,
    mapbox_center={"lat": 35.5, "lon": -79.0},
    margin={"r":0,"t":40,"l":0,"b":0},
    title="North Carolina Device Hot Spots (Hex Aggregation)"
)

fig.show()
```
